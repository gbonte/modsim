```python
import collections
import random

import simpy

```

```python
def mainsource(env, number, road):
    """Source generates customers randomly"""
    
    for i in range(number):
        c = mainstreet(env, 'CAR%02d' % i, road)
        env.process(c)
        t = random.expovariate(1.0/0.1 )
        yield env.timeout(t)

def secondarysource(env, number, road):
    """Source generates customers randomly"""
    
    for i in range(number):
        c = secondarystreet(env, 'car%02d' % i, road)
        env.process(c)
        t = random.expovariate(1.0/0.2 )
        yield env.timeout(t)        
        
def mainstreet(env,name,road):
    global green
    arrive=env.now
    print('%7.4f %s: Waiting for green' % (arrive,name))

    
    yield green
    with road.request() as req:
        yield req
        yield green
        print('%7.4f %s: Now is green' % (env.now,name))
        
        yield env.timeout(0.1)
            # We got to the counter
        print('%7.4f %s: Crossed ' % (env.now, name))

      
        

def secondarystreet(env,name,road):
    global green2
    global waitingtime
    arrive=env.now
    print('%7.4f %s: Waiting for green' % (arrive,name))
    yield green2
    with road.request() as req:
        yield req
        yield green2
        print('%7.4f %s: Now is green' % (env.now,name))
        
        yield env.timeout(0.1)
        
        crossingtime=env.now
        print('%7.4f %s: Crossed ' % (crossingtime, name))
        waitingtime=waitingtime+[crossingtime-arrive]

        

```

```python


def change_light(env):
    global green
    global green2
    
    green2.succeed()
    light="red"
    print('%7.4f Set mainstreet light to %s' % (env.now,light))
    while True:  
        if (light=="red"):
            yield env.timeout(0.25)
            light="green"
            green.succeed()
            green2=env.event()
        else:
            yield env.timeout(0.75)
            light="red"
            green=env.event()
            green2.succeed()
        print('%7.4f --------- Set mainstreet light to %s -------' % (env.now,light))
    


print('Traffic light')
SIM_TIME = 120  # Simulate until
env = simpy.Environment()

green = env.event()
green2 = env.event()

waitingtime=[]
WAITINGtime=[]
crossroad = simpy.Resource(env, capacity=2)
env.process(change_light(env))
env.process(mainsource(env, 1000, crossroad))
env.process(secondarysource(env, 1000, crossroad))

env.run(until=SIM_TIME)        


# Analysis/results
```
